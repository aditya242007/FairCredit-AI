# FairCredit AI — Data Audit

## Objective
The purpose of this notebook is to check the quality and structure of the credit card default dataset before applying machine learning.

### We will:
1. Load the dataset.
2. Check the number of rows and columns.
3. Understand column names and data types.
4. Check missing values and duplicate records.
5. Check the target variable distribution.
6. Identify basic data quality issues.

**Dataset:** Default of Credit Card Clients  
**Target column:** `default payment next month`

```python
import pandas as pd
```

```python
# Load the dataset
# Make sure the Excel file is inside the project's data folder.
df = pd.read_excel(
    "../data/default of credit card clients.xls",
    header=1
)

# Display the first five rows
display(df.head())
```

```python
# Display the number of rows and columns
print("Dataset shape:", df.shape)
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
```

```python
# Display all column names
print("Column names:")
print(df.columns.tolist())
```

```python
# Display column data types and non-null counts
df.info()
```

```python
# Count missing values in each column
print("Missing values in each column:")
print(df.isnull().sum())

# Count total missing values
print("\nTotal missing values:", df.isnull().sum().sum())
```

```python
# Check duplicate rows
print("Duplicate rows:", df.duplicated().sum())

# Check duplicate customer IDs
print("Duplicate IDs:", df["ID"].duplicated().sum())
```

```python
# Define the target column
target = "default payment next month"

# Count each target value
print("Target value counts:")
print(df[target].value_counts().sort_index())
```

```python
# Calculate the percentage of each target class
target_percentage = df[target].value_counts(normalize=True) * 100

print("Target percentages:")
print(target_percentage.sort_index().round(2))
```

```python
# Check unique values in selected columns
print("SEX values:", sorted(df["SEX"].unique()))
print("EDUCATION values:", sorted(df["EDUCATION"].unique()))
print("MARRIAGE values:", sorted(df["MARRIAGE"].unique()))

print("Target values:", sorted(df[target].unique()))
```

```python
# Check whether the target contains values other than 0 and 1
invalid_target = df[~df[target].isin([0, 1])]

print("Invalid target records:", len(invalid_target))
```

```python
# Display a short summary of the audit
print("========== DATA AUDIT SUMMARY ==========")

print("Total records:", df.shape[0])
print("Total columns:", df.shape[1])
print("Total missing values:", df.isnull().sum().sum())
print("Duplicate rows:", df.duplicated().sum())
print("Duplicate IDs:", df["ID"].duplicated().sum())
print("Invalid target records:", len(invalid_target))

print("\nTarget distribution:")
print(df[target].value_counts().sort_index())

print("\nData audit completed.")
```
